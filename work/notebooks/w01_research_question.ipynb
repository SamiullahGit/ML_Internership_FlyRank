## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Search question: Among pages with real demand, which should a reviewer look at first for refresh?
Decision improved: how a limited editorial team allocates its review time across ~30,000 pages.
Unit of analysis (grain): one content item (one page), deduplicated by content_id.
Who acts on it: a content editor/reviewer, who pulls the top-N ranked pages into the refresh workflow each sprint (N = the team's review capacity).
Output: a ranked review queue — a score and reason codes per page — not a yes/no flag.
Cost of a wrong recommendation: a false positive burns scarce editor hours refreshing a page that was fine; a false negative leaves a high-demand page quietly losing traffic. The missed high-demand decliner is the more expensive error, which is why I measure Precision@K rather than accuracy.
Why this needs data/ML, not a spreadsheet sort: the positive class isn't rare — over half of pages are declining — which is exactly why a binary flag fails: it names ~16,000 pages and a small team can act on a few dozen. The task is to rank those candidates so the highest-impact ones rise to the top, weighted by exposure. I also only have observed signals — FlyRank's own decision flags are deliberately excluded, so the model has to find signal instead of copying the product's answer. The model is one component: the starter blends it with a transparent baseline and attaches human-readable reason codes.
